## Two sample Z-test example (credit to Emma Ding)
### Scenario: 
Example from https://online.stat.psu.edu/stat415/lesson/9/9.4
n1= 605, n2 = 195, X1 = 351, X2=41

Null hypothesis $p_1-p_2 = 0$
Alternative hypothesis $p_1-p_2\neq 0$. Assuming null hypothesis is true $p_1 = p_2$ so we can infer $\hat{p_1} \sim \mathcal{N}(p,pq/n_1)$ and  $\hat{p_2} \sim \mathcal{N}(p,pq/n_2)$. So $\hat{p}_1-\hat{p}_2 \sim \mathcal{N}(0, pq(1/n_1+1/n_2))$ (Var(X1-X2) = Var(X1)+Var(X2), if two are independent)
Then, 
$$Z=\dfrac{(\hat{p}_1-\hat{p}_2)-0}{\sqrt{\hat{p}(1-\hat{p})\left(\dfrac{1}{n_1}+\dfrac{1}{n_2}\right)}},$$
where we estimate $\hat{p} = \frac{X_1+X_2}{n_1+n_2}$

```python
import scipy as sp
import numpy as np

sample_size_1= 605
sample_size_2 = 195
clicks_1 = 351
clicks_2 = 41
# Calculate the sample proportions
p_hat_1 = clicks_1 / sample_size_1 # p_hat_1~ N(p, pq/sample_size_1)
p_hat_2 = clicks_2 / sample_size_2 # p_hat_2~ N(p, pq/sample_size_2)
p_hat_tot = (clicks_1 + clicks_2) / (sample_size_1 + sample_size_2) # total sample proportion
print(f"Sample proportion 1: {p_hat_1: .3f}")
print(f"Sample proportion 2: {p_hat_2: .3f}")
print(f"Total sample proportion: {p_hat_tot: .3f}")

#null hypothesis: p_1 - p_2 = 0
#alternative hypothesis: p_1 - p_2 != 0
mu_diff = 0 # null hypothesis proportion difference
sigma_diff_pooled = np.sqrt(p_hat_tot * (1 - p_hat_tot) * (1/sample_size_1 + 1/sample_size_2)) # pooled standard deviation
sigma_diff_unpooled = np.sqrt(p_hat_1 * (1 - p_hat_1) / sample_size_1 + p_hat_2 * (1 - p_hat_2) / sample_size_2) # unpooled standard deviation
alpha = 0.05
z_score = (p_hat_1-p_hat_2-mu_diff) / sigma_diff_pooled # calculate z-score
z_crit = sp.stats.norm.ppf(1 - alpha/2) # critical value for one-tailed test
# Print the one proportion z test results
print(f"sigma_diff_pooled: {sigma_diff_pooled: .3f}")
print(f"sigma_diff_unpooled: {sigma_diff_unpooled: .3f}")
print(f"Z-score: {z_score: .3f}")
print(f"Critical value: {z_crit: .3f}")
if z_score > z_crit:    
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

### Confidence Interval Analysis
For large random samples, a $(1-\alpha)100\%$ confidence interval for a population proportion $p$ is
$$\hat{p}-z_{\alpha/2} \sigma_{\hat{p}}\leq p \leq \hat{p}+z_{\alpha/2}\sigma_{\hat{p}},$$
for $$\sigma_{\hat{p}} = \sqrt{\left(\dfrac{p_1q_q}{n_1}+\dfrac{p_2q_2}{n_2}\right)}$$ (i.e. using unpooled proportion)
where $z_{\alpha/2}$ satisfies $P(Z\geq z_{\alpha/2})=\alpha/2$ and $-z_{\alpha/2}$ satisfies $P(Z\leq -z_{\alpha/2})=\alpha/2$. Note that, we are essentially assuming:
$$P \left[-z_{\alpha/2}\leq \dfrac{p}{\sigma_{\hat{p}}} \leq z_{\alpha/2}\right] \approx 1-\alpha.$$

```python
# z_crit = sp.stats.norm.ppf(1 - alpha/2)

# Calculate the confidence interval
lower_bound = p_hat_1-p_hat_2 - z_crit * sigma_diff_unpooled
upper_bound = p_hat_1-p_hat_2 + z_crit * sigma_diff_unpooled
print(f"95% Confidence interval: ({lower_bound:.3f}, {upper_bound:.3f})")
```

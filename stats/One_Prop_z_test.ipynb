### Scenario: 
Suppose that we have an algorithm for Ad selection and we’d like to estimate the click-through rate
of users on the Ads selected by this algorithm. Given that we have access to 1000 users and the observed click-through rate is $\hat{p} = 0.2$. We set a significance level $\alpha =  5\%$.

### Statistical Testing (in Python)

```python
import scipy as sp
import numpy as np

p_hat = 0.2 #p_hat~ N(p, pq/n)
n = 1000 #sample size
p_null = 0.15
sigma = np.sqrt(p_null * (1 - p_null) / n)
alpha = 0.05
z_score = (p_hat - p_null) / sigma # calculate z-score
z_crit = sp.stats.norm.ppf(1 - alpha/2) # critical value for one-tailed test

# Print the one proportion z test results
print(f"Sample proportion (p_hat): {p_hat}")
print(f"z-score: {z_score:.3f}")
print(f"Critical value: {z_crit:.3f}")
if z_score > z_crit:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

### Confidence Interval Analysis
For large random samples, a $(1-\alpha)100\%$ confidence interval for a population proportion $p$ is
$$\hat{p}-z_{\alpha/2}\sqrt{\dfrac{\hat{p}(1-\hat{p})}{n}} \leq p \leq \hat{p}+z_{\alpha/2}\sqrt{\dfrac{\hat{p}(1-\hat{p})}{n}}, \text{for}$$
$z_{\alpha/2}$ being $P(Z\geq z_{\alpha/2})=\alpha/2$ and $-z_{\alpha/2}$ being $P(Z\leq -z_{\alpha/2})=\alpha/2$

```python
# z_crit = sp.stats.norm.ppf(1 - alpha/2)

# Calculate the confidence interval
sigma_phat = np.sqrt(p_hat * (1 - p_hat) / n)
lower_bound = p_hat - z_crit * sigma_phat
upper_bound = p_hat + z_crit * sigma_phat
print(f"95% Confidence interval: ({lower_bound:.3f}, {upper_bound:.3f})")
```
